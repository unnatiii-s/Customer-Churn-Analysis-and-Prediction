```python
import pandas as pd

# Load the dataset
df = pd.read_csv(r"C:\Users\unnati sutradhar\Downloads\internship\Telco_Customer_Churn_Dataset  (1).csv")

# Display the first few rows to understand the data
print(df.head())
```

```python
# Check for missing values
print(df.isnull().sum())

# Check data types to identify any hidden issues
print(df.info())
```

```python
# Convert 'TotalCharges' to numeric and handle errors
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# Drop rows with missing values in 'TotalCharges'
df = df.dropna()

# Check again to confirm
print(df.isnull().sum())
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

# Identify categorical columns
categorical_cols = df.select_dtypes(include=['object']).columns

# Exclude 'customerID' as it's a unique identifier
categorical_cols = [col for col in categorical_cols if col != 'customerID']

# Apply label encoding
for col in categorical_cols:
    df[col] = le.fit_transform(df[col])

print(df.head())
```

```python
# Apply one-hot encoding
df = pd.get_dummies(df, drop_first=True)

print(df.head())
```

```python
# Save the cleaned data for future use
df.to_csv('Cleaned_Telco_Customer_Churn.csv', index=False)
```
