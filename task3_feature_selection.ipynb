```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import mutual_info_classif, chi2
from sklearn.preprocessing import MinMaxScaler
from sklearn.feature_selection import SelectKBest

# Load the dataset (assuming it's already cleaned and split)
df = pd.read_csv(r'C:\Users\unnati sutradhar\Downloads\internship\saiketwork\Cleaned_Telco_Customer_Churn.csv')
X = df.drop(columns=['Churn'])  # Features
y = df['Churn']  # Target Variable
```

```python
# Correlation Analysis
plt.figure(figsize=(10, 6))
sns.heatmap(df.corr()['Churn'].sort_values(ascending=False).to_frame(), annot=True, cmap='coolwarm')
plt.title('Feature Correlation with Churn')
plt.show()
```

```python
# Feature Importance using RandomForest
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X, y)
importances = pd.DataFrame({'Feature': X.columns, 'Importance': rf.feature_importances_})
importances = importances.sort_values(by='Importance', ascending=False)

# Plot Top Features
plt.figure(figsize=(12, 6))
sns.barplot(x=importances['Importance'][:20], y=importances['Feature'][:20])
plt.title('Top 20 Features by Importance')
plt.show()
```

```python
# Chi-Square Test for Categorical Features
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
chi_scores, _ = chi2(X_scaled, y)
chi_results = pd.DataFrame({'Feature': X.columns, 'Chi2 Score': chi_scores})
chi_results = chi_results.sort_values(by='Chi2 Score', ascending=False)

# Display Top Features by Chi-Square Score
print(chi_results.head(20))
```

```python
# Mutual Information for Feature Selection
mi_scores = mutual_info_classif(X, y, random_state=42)
mi_results = pd.DataFrame({'Feature': X.columns, 'Mutual Information': mi_scores})
mi_results = mi_results.sort_values(by='Mutual Information', ascending=False)

# Display Top Features by Mutual Information Score
print(mi_results.head(20))
```
